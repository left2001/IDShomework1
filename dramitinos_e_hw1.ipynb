Answer b)In order to visualize if the PSA scores differ by sex, we need to have a grasp of the PSA scores' summary statistics if the sex is GIRL, BOY OR TOT.
So the perfect plot to do this is a boxplot.
 

```python
#Boxplot "PISA scores differ by sex"  
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,10))
sns.boxplot(data=data,x="sex",y="rating",palette='Set1')
plt.title("PISA scores by sex", fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Sex', fontsize=12)
plt.ylabel('PISA Score (Rating)', fontsize=12)
plt.tight_layout()
plt.show()
```

According to the boxplot above, there is no significant difference between the scores given their sex, because their quantiles do not manifest major differences.
